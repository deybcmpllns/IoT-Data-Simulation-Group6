```python
import pandas as pd

# Retrieve all IoT records
data = []
total_records = contract.functions.getTotalRecords().call()

for i in range(total_records):
    # Call the correct function from your ABI
    record = contract.functions.getShipmentData(i).call()
    
    # record is a tuple mapping directly to the 12 ABI function outputs
    data.append({
        "shipment_id": record[0],
        "rfid_tag": record[1],
        "vehicle_id": record[2],
        "origin": record[3],
        "destination": record[4],
        "status": record[5],
        "cargo_type": record[6],
        "temperature": record[7],
        "delay_hours": record[8],
        "co2_emissions": record[9],
        "timestamp": record[10],     # blockchainTimestamp is index 10
        "recorded_by": record[11]     # recordedBy address is index 11
    })

# Convert to a DataFrame
df = pd.DataFrame(data)

# Convert blockchainTimestamp to a readable format
# (Only do this if total_records > 0 to prevent pandas errors on empty data)
if not df.empty:
    df["timestamp"] = pd.to_datetime(df["timestamp"], unit="s")

# Display first few records
print(df.head())
```

```python
import numpy as np

# 1. Clean and extract numeric values from the specific string columns
df["temperature"] = df["temperature"].str.extract(r'(-?\d+\.?\d*)').astype(float)
df["delay_hours"] = df["delay_hours"].str.extract(r'(\d+\.?\d*)').astype(float)
df["co2_emissions"] = df["co2_emissions"].str.extract(r'(\d+\.?\d*)').astype(float)

# 2. Handle missing values (fills NaN values resulting from empty strings or failed extractions)
df.fillna({
    "temperature": 0.0,
    "delay_hours": 0.0,
    "co2_emissions": 0.0,
    "status": "Unknown"
}, inplace=True)

# Display cleaned data showing your specific logistics features
print(df[["shipment_id", "status", "temperature", "delay_hours", "co2_emissions"]].head())
```

```python
# Save cleaned IoT data to a CSV file
df.to_csv("cleaned_iot_data.csv", index=False)


print("✅ Cleaned IoT data saved successfully as cleaned_iot_data.csv")
```
