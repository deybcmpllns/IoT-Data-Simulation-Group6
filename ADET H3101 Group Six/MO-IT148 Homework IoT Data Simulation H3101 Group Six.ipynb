```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

num_records = 100
data = []

for _ in range(num_records):
    record = {
        "timestamp": datetime.now() - timedelta(minutes=np.random.randint(0, 1440)),
        "shipment_id": f"SHP{np.random.randint(1000,9999)}",
        "rfid_tag": f"RFID{np.random.randint(100000,999999)}",   # RFID tag ID
        "vehicle_id": f"VEH{np.random.randint(100,999)}",
        "driver_id": f"DRV{np.random.randint(10,99)}",
        "origin": np.random.choice(["Manila", "Cebu", "Davao", "Clark", "Iloilo"]),
        "destination": np.random.choice(["Manila", "Cebu", "Davao", "Clark", "Iloilo"]),
        "distance_km": round(np.random.uniform(50, 1000), 1),
        "scheduled_delivery": datetime.now() + timedelta(hours=np.random.randint(1,72)),
        "actual_delivery": datetime.now() + timedelta(hours=np.random.randint(1,72)),
        "status": np.random.choice(["In Transit", "Delivered", "Delayed", "Cancelled"]),
        "cargo_weight_kg": round(np.random.uniform(100, 5000), 1),
        "cargo_type": np.random.choice(["Electronics", "Food", "Clothing", "Furniture", "Pharmaceuticals"]),
        "fuel_consumed_l": round(np.random.uniform(10, 300), 1),
        "avg_speed_kmh": round(np.random.uniform(40, 100), 1),
        "temperature_c": round(np.random.uniform(15, 35), 1),
        "transport_cost_usd": round(np.random.uniform(50, 2000), 2),
        "co2_emissions_kg": round(np.random.uniform(20, 500), 2)
    }
    data.append(record)
```

```python
df = pd.DataFrame(data)
```

```python
df["delay_hours"] = (df["actual_delivery"] - df["scheduled_delivery"]).dt.total_seconds() / 3600
```

```python
df.to_csv("logistics_data.csv", index=False)
df.to_json("logistics_data.json", orient="records")
```
